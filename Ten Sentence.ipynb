# Introduction

This study examines whether flipper length is associated with body mass among penguins in the Seaborn penguins dataset.
Understanding this relationship can help describe how different measures of penguin body size vary together.

# Literature Review / Theoretical Frame

As an exploratory premise, larger penguins may have longer flippers because different body dimensions tend to scale together.
However, differences between species may also contribute to an overall association between flipper length and body mass.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from IPython.display import display, Markdown

# Load the data
df = sns.load_dataset("penguins")

# Keep complete pairs for the two variables
df = df[["flipper_length_mm", "body_mass_g"]].dropna()

n = len(df)
```

```python
method_text = (
    f"We analyzed {n} complete observations from the Seaborn "
    "penguins dataset, measuring flipper length in millimeters "
    "and body mass in grams. "
    "We calculated Pearson's correlation coefficient to describe "
    "the linear association, with the null hypothesis that the "
    "population correlation is zero and the alternative hypothesis "
    "that it is nonzero."
)

display(Markdown("# Method\n\n" + method_text))
```

```python
# Calculate the correlation matrix
corr_matrix = df.corr(method="pearson")
display(corr_matrix)

# Extract the coefficient using row and column labels
r = corr_matrix.loc["flipper_length_mm", "body_mass_g"]

direction = (
    "positive" if r > 0
    else "negative" if r < 0
    else "zero"
)

interpretation = (
    "penguins with longer flippers tended to have greater body mass"
    if r > 0
    else "penguins with longer flippers tended to have lower body mass"
    if r < 0
    else "there was no linear association in the sample"
)
```

```python
results_text = (
    f"The Pearson correlation between flipper length and body mass "
    f"was r = {r:.3f}, indicating a {direction} linear association. "
    f"In this sample, {interpretation}."
)

display(Markdown("# Results\n\n" + results_text))
```

# Discussion

The pooled association may partly reflect differences between species and does not establish a causal effect of flipper length on body mass.
Future research should examine within-species relationships to determine whether the overall pattern also appears among penguins of the same species.
