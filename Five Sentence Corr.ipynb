# Introduction

This study examines whether flipper length is associated with body mass among penguins in the Seaborn penguins dataset.

# Literature Review / Theoretical Frame

As an exploratory premise, larger penguins may have longer flippers, suggesting a positive relationship between flipper length and body mass.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import pearsonr
from IPython.display import display, Markdown

# Load data and keep observations with both measurements
df = sns.load_dataset("penguins")
df = df.dropna(
    subset=["flipper_length_mm", "body_mass_g"]
)

n = len(df)
```

```python
method_sentence = (
    f"Using {n} complete observations with flipper length "
    "measured in millimeters and body mass in grams, "
    "we conduct a two-sided Pearson correlation test "
    "of the null hypothesis that the population correlation "
    "is zero against the alternative hypothesis that it is nonzero."
)

display(Markdown("# Method\n\n" + method_sentence))
```

```python
# Calculate Pearson's correlation and p-value
r, pval = pearsonr(
    df["flipper_length_mm"],
    df["body_mass_g"]
)

# Prepare text for the results
p_text = "p < .001" if pval < 0.001 else f"p = {pval:.3f}"

direction = (
    "positive" if r > 0
    else "negative" if r < 0
    else "zero"
)

decision = (
    "reject" if pval < 0.05
    else "fail to reject"
)
```

```python
results_sentence = (
    f"The analysis yielded a {direction} Pearson correlation "
    f"of r = {r:.3f}, with {p_text}, leading us to "
    f"{decision} the null hypothesis at α = .05."
)

display(Markdown("# Results\n\n" + results_sentence))
```

# Discussion
The pooled correlation may partly reflect differences between penguin species, so future research should examine within-species relationships rather than interpret the association as causal.
