# Introduction
This study examines whether penguin species is associated with the island where penguins were observed in the Seaborn penguins dataset.

# Literature Review / Theoretical Frame
We consider sex as a possible correlate of smoking status without assuming that the two variables are necessarily associated.

# Method
Using 344 observations from the Seaborn penguins dataset, we conducted a chi-square test of independence, with the null hypothesis that penguin species and island are independent and the alternative hypothesis that they are associated.


# Results + Analysis
The chi-square test yielded χ² = 299.55, df = 4, and p < .001, leading us to reject the null hypothesis and conclude that penguin species and island are significantly associated.

# Discussion
The findings suggest that penguin species are unevenly distributed across the sampled islands, although these observational data cannot establish the causes of this association.
